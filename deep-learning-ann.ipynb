```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from tensorflow.keras import layers
```

```python
df = pd.read_csv('/kaggle/input/datasets/vivekyash/heart-failure-prediction-deep-learning-ann/heart_failure_clinical_records_dataset.csv')
df
```

```python
df.columns
```

```python
x = df[['age', 'anaemia', 'creatinine_phosphokinase', 'diabetes',
       'ejection_fraction', 'high_blood_pressure', 'platelets',
       'serum_creatinine', 'serum_sodium', 'sex', 'smoking', 'time']]
y = df[['DEATH_EVENT']]
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
x_min = x.min()
x_max = x.max()
x_scaled = (x - x_min)/(x_max - x_min + 1e-8)
x_scaled
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size = 0.25, random_state = 42 )
```

```python
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '-1'
```

```python
model = keras.Sequential([
    layers.Input(shape=(x_train.shape[1],)),
    layers.Dense(8, activation= 'relu'),
    layers.Dense(6, activation= 'relu'),
    layers.Dense(4, activation= 'relu'),
    layers.Dense(2, activation= 'relu'),
    layers.Dense(1, activation= 'linear')
])
```

```python
model.compile(optimizer='sgd', loss='BinaryCrossentropy', metrics = ['accuracy'])
```

```python
history = model.fit(
    x_train.values, y_train.values, validation_data=(x_test.values,y_test.values),
    epochs = 100,batch_size = 4,verbose = 1
)
```

```python
highest_train_acc = max(history.history['accuracy'])
highest_train_acc
```
